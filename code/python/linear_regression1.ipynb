# Linear regression 

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
from sklearn.preprocessing import OneHotEncoder
```

```python
df = pd.read_csv('..//data//Human_Activity_Recognition_Using_Smartphones_Data-1.csv')
```

```python
X = df.drop('Activity', axis=1)
y = df['Activity']
```

```python
encoder = OneHotEncoder(sparse=False)
y= encoder.fit_transform(df[['Activity']])
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

# Regularization

## Ridge Regression (L2 Regularization) 

```python
from sklearn.linear_model import Ridge
```

```python
ridge_reg = Ridge(alpha=1.0)
```

```python
ridge_reg.fit(X_train, y_train)
```

```python
y_pred_ridge = ridge_reg.predict(X_test)
```

```python
mse_ridge = mean_squared_error(y_test, y_pred_ridge, multioutput='raw_values')
print(f'The mse with L2 regularization is: {mse_ridge}%')
```

## Lasso Regression (L1 Regularization)

```python
from sklearn.linear_model import Lasso
```

```python
lasso_reg = Lasso(alpha=0.001)
```

```python
lasso_reg.fit(X_train, y_train)
```

```python
y_pred_lasso = lasso_reg.predict(X_test)
```

```python
mse_lasso = mean_squared_error(y_test, y_pred_lasso, multioutput='raw_values')
print(f'The mse with L1 regularization is: {mse_lasso}%')
```

## Elastic Net (L1 + L2 combined)

```python
from sklearn.linear_model import ElasticNet
```

```python
elastic_net = ElasticNet(alpha=0.01, l1_ratio=0.2)
```

```python
elastic_net.fit(X_train, y_train)
```

```python
y_pred_elastic = elastic_net.predict(X_test)
```

```python
mse_elastic_net = mean_squared_error(y_test, y_pred_elastic, multioutput='raw_values')
print(f'The mse with elastic net is: {mse_elastic_net}%')
```

## LARS Lasso (Least Absolute Shrinkage and Selection Operator)

```python
from sklearn.linear_model import Lars
```

```python
lars_lasso = Lars(eps=25, normalize=False)
```

```python
lars_lasso.fit(X_train, y_train)
```

```python
y_pred_lars_lasso = lars_lasso.predict(X_test)
```

```python
mse_lars_lasso = mean_squared_error(y_test, y_pred_lars_lasso, multioutput='raw_values')
print(f'The mse with elastic net is: {mse_lars_lasso}%')
```

# Variations

## Polynomial regression

```python
from sklearn.preprocessing import PolynomialFeatures
```

```python
poly = PolynomialFeatures(degree=2)
```

```python
X_train_poly = poly.fit_transform(X_train)
```

```python
X_test_poly = poly.transform(X_test)
```

```python
poly_reg = LinearRegression()
```

```python
poly_reg.fit(X_train_poly, y_train)
```

```python
y_pred_poly = poly_reg.predict(X_test_poly)
```

```python
mse_poly = mean_squared_error(y_test, y_pred_poly, multioutput='raw_values')
print(f'The mse with polynomial regression is: {mse_poly}%')
```
