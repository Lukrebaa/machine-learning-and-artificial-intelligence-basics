# KNN

```python
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv('..//data//Human_Activity_Recognition_Using_Smartphones_Data-1.csv')
```

```python
X = df.drop('Activity', axis=1)
y = df['Activity']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

# Regularization

## Number of k

```python
from sklearn.model_selection import GridSearchCV
```

```python
param_grid = {'n_neighbors': range(1, 31)}
grid_search = GridSearchCV(KNeighborsClassifier(), param_grid, cv=5)
grid_search.fit(X_train, y_train)
```

```python
best_k = grid_search.best_params_['n_neighbors']
```

```python
knn_optimized_k = KNeighborsClassifier(n_neighbors=best_k)
knn_optimized_k.fit(X_train, y_train)
```

```python
y_pred_optimized_k = knn_optimized_k.predict(X_test)
```

```python
accuracy_optimized_k = accuracy_score(y_test, y_pred_optimized_k)
print("Best number of neighbors:", best_k)
print("Accuracy with optimized k:", accuracy_optimized_k*100)
```

## Distance metrics

### Manhattan

```python
knn_manhattan = KNeighborsClassifier(n_neighbors=5, metric='manhattan')
knn_manhattan.fit(X_train, y_train)
```

```python
y_pred_knn_manhattan = knn_manhattan.predict(X_test)
```

```python
accuracy_knn_manhattan = accuracy_score(y_test, y_pred_knn_manhattan)
print(f'The accuracy is: {accuracy_knn_manhattan*100}%')
```

### Euclidean

```python
knn_euclidean = KNeighborsClassifier(n_neighbors=5, metric='euclidean')
knn_euclidean.fit(X_train, y_train)
```

```python
y_pred_knn_euclidean = knn_euclidean.predict(X_test)
```

```python
accuracy_knn_euclidean = accuracy_score(y_test, y_pred_knn_euclidean)
print(f'The accuracy is: {accuracy_knn_euclidean*100}%')
```

## Feature scaling

```python
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv('..//data//Human_Activity_Recognition_Using_Smartphones_Data-1.csv')
X = df.drop('Activity', axis=1)
y = df['Activity']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
knn_scaled = KNeighborsClassifier(n_neighbors=1)
knn_scaled.fit(X_train_scaled, y_train)
```

```python
y_pred_scaled_knn = knn_scaled.predict(X_test_scaled)
```

```python
accuracy_scaled_knn = accuracy_score(y_test, y_pred_scaled_knn)
print(f'The accuracy is: {accuracy_scaled_knn*100}%')
```

## Dimensionality reduction

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components=0.95) 
X_train_pca = pca.fit_transform(X_train)
X_test_pca = pca.transform(X_test)
```

```python
knn_pca = KNeighborsClassifier(n_neighbors=5)
knn_pca.fit(X_train_pca, y_train)
```

```python
y_pred_knn_pca = knn_pca.predict(X_test_pca)
```

```python
accuracy_knn_pca = accuracy_score(y_test, y_pred_knn_pca)
print(f'The accuracy is: {accuracy_knn_pca*100}%')
```

# Variations

## Weighted KNN

```python
weighted_knn = KNeighborsClassifier(n_neighbors=5, weights='distance')
weighted_knn.fit(X_train, y_train)
```

```python
y_pred_weighted_knn = weighted_knn.predict(X_test)
```

```python
accuracy_weighted_knn = accuracy_score(y_test, y_pred_weighted_knn)
print(f'The accuracy is: {accuracy_weighted_knn*100}%')
```
