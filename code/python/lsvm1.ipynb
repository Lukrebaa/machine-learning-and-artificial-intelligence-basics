# Linear Support Vector Machine

```python
import pandas as pd
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv('..//data//Human_Activity_Recognition_Using_Smartphones_Data-1.csv')
```

```python
X = df.drop('Activity', axis=1)
y = df['Activity']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

# Regularization

## C regularization

```python
C = [0.01, 0.1, 1, 10, 100]
```

```python
accuracy_scores = {}
```

```python
for Cval in C:
    svm_model = SVC(C=Cval, kernel='linear', random_state=42)
    svm_model.fit(X_train, y_train)
    y_pred_svm = svm_model.predict(X_test)
    accuracy = accuracy_score(y_test, y_pred_svm)
    accuracy_scores[Cval] = accuracy
    print(f'The accuracy with {Cval} is: {accuracy_scores[Cval]*100}%')
```

# Kernel tricks

## Linear kernel

```python
svm_linear_kernel = SVC(kernel='linear', random_state=42)
svm_linear_kernel.fit(X_train, y_train)
```

```python
y_pred_svm_linear_kernel = svm_linear_kernel.predict(X_test)
```

```python
accuracy_svm_linear_kernel = accuracy_score(y_test, y_pred_svm_linear_kernel)
print(f'The accuracy is: {accuracy_svm_linear_kernel*100}%')
```

```python
accuracy_svm_poly_kernel = accuracy_score(y_test, y_pred_svm_poly_kernel)
print(f'The accuracy is: {accuracy_svm_poly_kernel*100}%')
```

```python
y_pred_svm_rbf_kernel = svm_rbf_kernel.predict(X_test)
```

## Support Vector Regression

```python
from sklearn.metrics import mean_squared_error
from sklearn.preprocessing import OneHotEncoder
from sklearn.svm import SVR
```

```python
encoder = OneHotEncoder(sparse=False)
y_oh= encoder.fit_transform(df[['Activity']])
```

```python
X_train_svr_oh, X_test_svr_oh, y_train_svr_oh, y_test_svr_oh = train_test_split(X, y_oh, test_size=0.3, random_state=42)
```

```python
svr_models = [SVR() for _ in range(y_oh.shape[1])]
mse_svr_oh = []
```

```python
for i in range(y_oh.shape[1]):
    svr_models[i].fit(X_train_svr_oh, y_train_svr_oh[:, i])
    y_pred_svr_oh = svr_models[i].predict(X_test_svr_oh)
    mse = mean_squared_error(y_test_svr_oh[:, i], y_pred_svr_oh)
    mse_svr_oh.append(mse)
    print(f'The accuracy with {i} is: {mse_svr_oh[i]}%')
```

## Kernel approximation

```python
from sklearn.kernel_approximation import RBFSampler
from sklearn.linear_model import SGDClassifier
```

```python
rbf_feature = RBFSampler(gamma=0.001, random_state=42)
X_train_kernel_approx = rbf_feature.fit_transform(X_train)
X_test_kernel_approx = rbf_feature.transform(X_test)
```

```python
svm_approx = SGDClassifier(max_iter=1000, random_state=42)
svm_approx.fit(X_train_kernel_approx, y_train)
```

```python
y_pred_svm_approx = svm_approx.predict(X_test_kernel_approx)
```

```python
accuracy_svm_approx = accuracy_score(y_test, y_pred_svm_approx)
print(f'The accuracy is: {accuracy_svm_approx*100}%')
```
